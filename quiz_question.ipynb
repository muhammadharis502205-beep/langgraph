```python
from langgraph.graph import StateGraph,START,END
from langchain_mistralai import ChatMistralAI
from dotenv import load_dotenv
from pydantic import BaseModel
```

```python
load_dotenv(override=True)
```

```python
llm=ChatMistralAI(
    model="open-mistral-7b"
)
```

```python
class state(BaseModel):
    topic:str
    pessage:str
    key_points:str
    questions:str
```

```python
def Pessage_node(state:state):
    response=llm.invoke(f"Give me short pessage about : {state.topic}")
    return {"pessage":response.content}
```

```python
def Key_points_node(state:state):
    response=llm.invoke(f"Give me five key points about this pessage : {state.pessage}")
    return {"key_points":response.content}
```

```python
def Question_node(state:state):
    response=llm.invoke(f"Give me simple five quiz question base on this key points : {state.key_points}")
    return {"questions":response.content}
```

```python
graph_builder=StateGraph(state)
```

```python
graph_builder.add_node("Pessage_node",Pessage_node)
graph_builder.add_node("Key_points_node",Key_points_node)
graph_builder.add_node("Question_node",Question_node)
```

```python
graph_builder.add_edge(START,"Pessage_node")
graph_builder.add_edge("Pessage_node","Key_points_node")
graph_builder.add_edge("Key_points_node","Question_node")
graph_builder.add_edge("Question_node",END)
```

```python
graph=graph_builder.compile()
```

```python
initial_state=state(topic="Autonomius ai agents",pessage="",key_points="",questions="")
```

```python
result=graph.invoke(initial_state)
```
